```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv("datasets/car-data.csv", usecols=["name","seller_type","transmission","owner","fuel"])
```

```python
df.head()
```

```python
df = df.drop(["name"], axis=1)
```

```python
df
```

Using Pandas

```python
# How many types of values in column
df["transmission"].value_counts()
```

```python
df["transmission"] = df["transmission"].map({"Manual": 0, "Automatic": 1})
df.sample(5)
```

```python
df["seller_type"] = df["seller_type"].map({"Individual": 0, "Dealer": 1,"Trustmark Dealer": 2})
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop("fuel", axis=1), df["fuel"], test_size=0.33, random_state=42)
```

```python
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
```

Appling Label Encoder

```python
le = LabelEncoder()
```

```python
y_train_scaled =le.fit_transform(y_train)
y_test_scaled = le.transform(y_test)
```

```python
df["fuel"].value_counts()
```

```python
oe = OrdinalEncoder()
X_train_scaled =oe.fit_transform(X_train)
X_test_scaled = oe.transform(X_test)
```
