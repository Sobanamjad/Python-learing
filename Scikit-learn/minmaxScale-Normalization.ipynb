```python
import pandas as pd
```

```python
df = pd.read_csv("datasets/car-data.csv")
```

```python
df
```

```python
df = df[["selling_price", "year", "km_driven"]]
```

```python
df
```

Appling Normalizaqtion : Min Max Scaler

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
# Creating object

scaler = MinMaxScaler()
```

```python
# Data Transform 
scald_df = scaler.fit_transform(df)
```

```python
# Numpy Array 
scald_df
```

```python
# Making Data Frame
scald_df = pd.DataFrame(data=scald_df, columns=df.columns)
scald_df
```

```python
# Train Test Split
from sklearn.model_selection import train_test_split
```

```python
x = df.drop("selling_price", axis=1)
y = df["selling_price"]
x
y
```

```python
x_train , x_test, y_train, y_test = train_test_split( x, y, test_size=0.33, random_state=42)
```

Featured-Scalling: Normalization: MinMax Scaler

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled, columns=x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled, columns=x_test.columns)
```
